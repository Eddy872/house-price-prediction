# House Price Prediction (Kaggle)

Ce projet vise à prédire le prix de logements à partir du dataset Kaggle [House Prices - Advanced Regression Techniques](https://www.kaggle.com/c/house-prices-advanced-regression-techniques).

---

## Objectifs
- Prétraiter et nettoyer les données
- Appliquer un modèle de régression linéaire
- Évaluer les performances avec la RMSE
- Visualiser les résultats

---

## Structure du projet
- `house_prediction.ipynb` : notebook principal avec tout le code et les explications
- `data/Housing.csv` : données brutes ( à télécharger manuellement depuis Kaggle)

---

## Librairies utilisées
- Python 3.x
- pandas
- numpy
- matplotlib
- seaborn
- scikit-learn
